```python
def col_filter(df):
    metrics = [
        "accuracy",
        "balanced_accuracy",
        "f1_weighted",
        "roc_auc",
        "log_loss",
    ]

    cols = [
        c for c in df.columns
        if any(c.startswith(m + "_") for m in metrics)
    ]

    return df.loc[["XGBoost"], cols]
```

```python
import numpy as np
import pandas as pd

# ----------------------------
# Formatting: no scientific notation
# ----------------------------
pd.set_option("display.float_format", lambda x: f"{x:.6f}")  # only used if raw floats appear
pd.set_option("display.max_rows", 500)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)

METRICS = [
    ("accuracy", "accuracy_mean", "accuracy_std"),
    ("balanced_accuracy", "balanced_accuracy_mean", "balanced_accuracy_std"),
    ("F1 (weighted)", "f1_weighted_mean", "f1_weighted_std"),
    ("ROC AUC", "roc_auc_mean", "roc_auc_std"),
    ("log-loss", "log_loss_mean", "log_loss_std"),
]

def _parse_key(key: str):
    parts = key.split()
    dataset = parts[0]
    gen = parts[1]
    tau = None
    if len(parts) >= 3:
        try:
            tau = float(parts[2])
        except ValueError:
            tau = None
    return dataset, gen, tau

def _get_row(df: pd.DataFrame):
    # your frames are 1-row with index 'XGBoost'
    if df.shape[0] == 1:
        return df.iloc[0]
    if "XGBoost" in df.index:
        return df.loc["XGBoost"]
    return df.iloc[0]

def _fmt(mean, std, nd_mean=3, nd_std=3):
    # fixed-point formatting, never scientific notation
    return f"{float(mean):.{nd_mean}f} ± {float(std):.{nd_std}f}"

def build_notebook_summary_table(trtr_all: dict, tstr_all: dict, nd_mean=3, nd_std=3) -> pd.DataFrame:
    """
    Output columns:
      Dataset | Generator | Metric | REAL | Baseline | Min_{tau_ANY} | Max_{tau_ANY}
    All cells are strings "m ± s" with fixed decimals (no scientific notation).
    """

    # collect tau keys per (dataset, gen)
    tstr_keys_by_pair = {}
    for k in tstr_all.keys():
        ds, gen, tau = _parse_key(k)
        tstr_keys_by_pair.setdefault((ds, gen), []).append((tau, k))

    rows = []
    for (ds, gen), tau_keys in sorted(tstr_keys_by_pair.items()):
        base_key = f"{ds} {gen}"

        if base_key not in trtr_all:
            raise KeyError(f"Missing REAL key in trtr_all: '{base_key}'")
        if base_key not in tstr_all:
            raise KeyError(f"Missing Baseline key in tstr_all: '{base_key}'")

        real_row = _get_row(trtr_all[base_key])
        base_row = _get_row(tstr_all[base_key])

        tau_entries = [(tau, k) for (tau, k) in tau_keys if tau is not None]
        tau_rows = [(tau, _get_row(tstr_all[k])) for (tau, k) in tau_entries]

        for metric_label, mean_col, std_col in METRICS:
            real_cell = _fmt(real_row[mean_col], real_row[std_col], nd_mean, nd_std)
            base_cell = _fmt(base_row[mean_col], base_row[std_col], nd_mean, nd_std)

            if not tau_rows:
                min_cell = ""
                max_cell = ""
            else:
                means = np.array([float(r[mean_col]) for _, r in tau_rows], dtype=float)
                imin = int(np.argmin(means))
                imax = int(np.argmax(means))
                rmin = tau_rows[imin][1]
                rmax = tau_rows[imax][1]
                min_cell = _fmt(rmin[mean_col], rmin[std_col], nd_mean, nd_std)
                max_cell = _fmt(rmax[mean_col], rmax[std_col], nd_mean, nd_std)

            rows.append({
                "Dataset": ds.capitalize(),
                "Generator": gen,
                "Metric": metric_label,
                "REAL": real_cell,
                "Baseline": base_cell,
                "Min_{tau_ANY}": min_cell,
                "Max_{tau_ANY}": max_cell,
            })

    return pd.DataFrame(rows)

def merge_dicts(*dicts):
    out = {}
    for d in dicts:
        out.update(d)
    return out

# ----------------------------
# Example (plug in your dict variables)
# ----------------------------
trtr_all = merge_dicts(credit_ctgan_trtr, credit_tvae_trtr, adult_ctgan_trtr, adult_tvae_trtr, cardio_ctgan_trtr, cardio_tvae_trtr)
tstr_all = merge_dicts(credit_ctgan_tstr, credit_tvae_tstr, adult_ctgan_tstr, adult_tvae_tstr, cardio_ctgan_tstr, cardio_tvae_tstr)

df = build_notebook_summary_table(trtr_all, tstr_all, nd_mean=3, nd_std=8)


df
```
