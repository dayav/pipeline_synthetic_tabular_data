## Distance Based membership inference attack

### distance based measure based on small data

```python
# Worked examples for distance-based MIA functions on a small Adult-like toy dataset.
# This cell defines the functions and runs illustrative examples with printed outputs.

import numpy as np
import pandas as pd
from math import sqrt

# ----------------------------
# Utilities: type handling and distances
# ----------------------------
def _split_cols(df: pd.DataFrame):
    num_cols = df.select_dtypes(include=['number']).columns.tolist()
    cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    # treat bool as categorical
    for c in df.select_dtypes(include=['bool']).columns:
        if c not in cat_cols:
            cat_cols.append(c)
    # ensure stable order
    return num_cols, cat_cols

def _get_scaler(df: pd.DataFrame, num_cols):
    # min/max for numeric scaling; avoid zero range
    return {c: (float(df[c].min()), float(df[c].max())) for c in num_cols}

def _row_distance(row1, row2, num_cols, cat_cols, scaler, metric: str):
    # metric: 'gower' (avg of feature distances), 'heom' (euclidean over mixed), 'euclidean' (euclidean over normalized features)
    # NaNs in categoricals count as mismatch of 1.0. NaNs in numerics treated as max distance (1.0) as well.
    metric = metric.lower()
    if metric not in ('gower', 'heom', 'euclidean'):
        raise ValueError("Unsupported metric; choose 'gower', 'heom', or 'euclidean'")
    # accumulate
    if metric == 'heom':
        s = 0.0
        # numerics
        for c in num_cols:
            lo, hi = scaler[c]
            rng = hi - lo if hi != lo else 1.0
            x = row1[c]; y = row2[c]
            if pd.isna(x) or pd.isna(y):
                d = 1.0
            else:
                d = (float(x) - float(y)) / rng
            s += d * d
        # categoricals
        for c in cat_cols:
            x = row1[c]; y = row2[c]
            diff = 1.0 if (pd.isna(x) or pd.isna(y) or x != y) else 0.0
            s += diff * diff  # diff is 0 or 1
        return sqrt(s)
    else:
        # accumulate either sum of abs (gower) or sum of squared (euclidean), then normalize
        s = 0.0
        cnt = 0
        for c in num_cols:
            lo, hi = scaler[c]
            rng = hi - lo if hi != lo else 1.0
            x = row1[c]; y = row2[c]
            if pd.isna(x) or pd.isna(y):
                d = 1.0  # conservative
            else:
                d = abs(float(x) - float(y)) / rng
            if metric == 'euclidean':
                s += d * d
            else:
                s += d
            cnt += 1
        for c in cat_cols:
            x = row1[c]; y = row2[c]
            diff = 1.0 if (pd.isna(x) or pd.isna(y) or x != y) else 0.0
            if metric == 'euclidean':
                s += diff * diff
            else:
                s += diff
            cnt += 1
        if metric == 'euclidean':
            return sqrt(s)
        else:
            return s / (cnt if cnt > 0 else 1.0)

# ----------------------------
# Core functions
# ----------------------------
def dcr_membership_score(real_df: pd.DataFrame, candidate_rows: pd.DataFrame, 
                         distance: str = 'gower', scaler=None, k: int = 1):
    """
    Return per-row nearest-real distance (DCR). If k>1, return the mean distance to the k nearest real neighbors.
    Lower values => more likely member (closer to some real training record).
    """
    num_cols, cat_cols = _split_cols(real_df)
    if scaler is None:
        scaler = _get_scaler(real_df, num_cols)
    real_df = real_df.reset_index(drop=True)
    cand = candidate_rows.reset_index(drop=True)
    out = []
    for i, r in cand.iterrows():
        # compute distances to all real rows
        dists = []
        for j, rr in real_df.iterrows():
            d = _row_distance(r, rr, num_cols, cat_cols, scaler, distance)
            dists.append(d)
        dists.sort()
        k_eff = min(k, len(dists))
        if k_eff <= 0:
            out.append(np.nan)
        else:
            out.append(float(np.mean(dists[:k_eff])))
    return np.array(out)

def k_nn_mia_score(real_df: pd.DataFrame, synth_df: pd.DataFrame, 
                   distance: str = 'gower', k: int = 5, agg: str = 'mean'):
    """
    Return per-sample k-NN based membership scores (lower=more likely member).
    agg='mean' returns mean of k-NN distances; agg='max' returns the k-th neighbor distance.
    """
    num_cols, cat_cols = _split_cols(real_df)
    scaler = _get_scaler(real_df, num_cols)
    real_df = real_df.reset_index(drop=True)
    synth_df = synth_df.reset_index(drop=True)
    scores = []
    for i, s in synth_df.iterrows():
        dists = []
        for j, rr in real_df.iterrows():
            dists.append(_row_distance(s, rr, num_cols, cat_cols, scaler, distance))
        dists.sort()
        if len(dists) == 0:
            scores.append(np.nan)
            continue
        if k <= 1:
            scores.append(float(dists[0]))
        else:
            k_eff = min(k, len(dists))
            if agg == 'mean':
                scores.append(float(np.mean(dists[:k_eff])))
            elif agg == 'max':
                scores.append(float(dists[k_eff-1]))
            else:
                raise ValueError("agg must be 'mean' or 'max'")
    return pd.Series(scores, index=synth_df.index, name=f'{k}-NN_{agg}_{distance}')

def nn_classifier_mia(real_df: pd.DataFrame, holdout_df: pd.DataFrame, distance: str = 'gower', protocol: str = 'holdout'):
    """
    1-NN adversary on pooled (members=real_df, non-members=holdout_df) with leave-one-out for fair evaluation.
    Returns: dict with accuracy, auc (pairwise probability), and per-row details.
    """
    assert protocol in ('holdout', 'pooled-loo'), "protocol must be 'holdout' or 'pooled-loo'"
    # Build combined dataset with labels: 1 for member (train), 0 for non-member (holdout)
    A = real_df.copy().reset_index(drop=True)
    B = holdout_df.copy().reset_index(drop=True)
    A['__label__'] = 1
    B['__label__'] = 0
    combined = pd.concat([A, B], ignore_index=True)
    num_cols, cat_cols = _split_cols(combined.drop(columns=['__label__']))
    scaler = _get_scaler(combined, num_cols)
    # For each point, compute distance to nearest member (excluding self if member) and nearest non-member
    nearest_member = []
    nearest_nonmem = []
    pred = []
    for i, x in combined.iterrows():
        # nearest member
        d_min_mem = np.inf
        d_min_non = np.inf
        for j, y in combined.iterrows():
            if i == j:
                continue
            d = _row_distance(x, y, num_cols, cat_cols, scaler, distance)
            if y['__label__'] == 1 and d < d_min_mem:
                d_min_mem = d
            if y['__label__'] == 0 and d < d_min_non:
                d_min_non = d
        nearest_member.append(d_min_mem)
        nearest_nonmem.append(d_min_non)
        # predict membership by whichever nearest is closer: smaller distance wins
        pred.append(1 if d_min_mem <= d_min_non else 0)
    combined['d_nn_member'] = nearest_member
    combined['d_nn_nonmember'] = nearest_nonmem
    combined['pred'] = pred
    # Accuracy
    y_true = combined['__label__'].values
    y_pred = combined['pred'].values
    acc = float(np.mean(y_true == y_pred))
    # Continuous score for AUC: larger means more likely member
    scores = (combined['d_nn_nonmember'] - combined['d_nn_member']).values  # positive -> closer to member side
    # Manual pairwise AUC (handles ties as 0.5 credit)
    pos_scores = scores[y_true == 1]
    neg_scores = scores[y_true == 0]
    if len(pos_scores) == 0 or len(neg_scores) == 0:
        auc = np.nan
    else:
        wins = 0.0
        total = 0.0
        for ps in pos_scores:
            for ns in neg_scores:
                if ps > ns:
                    wins += 1.0
                elif ps == ns:
                    wins += 0.5
                total += 1.0
        auc = float(wins / total) if total > 0 else np.nan
    return {
        'accuracy': acc,
        'auc': auc,
        'details': combined.drop(columns=['__label__'])
    }

def local_density_ratio(real_df: pd.DataFrame, synth_df: pd.DataFrame, k: int = 5, distance: str = 'gower'):
    """
    Estimate density(real)/density(synth) via k-NN radii:
    ratio_i = (d_k_real(i) + eps) / (d_k_synth(i) + eps)
    Where d_k_real(i) is distance from real point i to its k-th nearest other real point (LOO),
    and d_k_synth(i) is distance from real point i to its k-th nearest synthetic neighbor.
    Interpretation:
      ratio < 1  -> synth is denser around i than real: potential overfitting risk for record i.
      ratio ~ 1  -> similar local density: neutral.
      ratio > 1  -> real denser than synth: generator under-represents that region (utility issue).
    """
    eps = 1e-9
    real_df = real_df.reset_index(drop=True)
    synth_df = synth_df.reset_index(drop=True)
    num_cols, cat_cols = _split_cols(real_df)
    scaler = _get_scaler(real_df, num_cols)
    ratios = []
    d_real_k_list = []
    d_synth_k_list = []
    for i, x in real_df.iterrows():
        # distances to other reals (leave-one-out)
        dists_real = []
        for j, y in real_df.iterrows():
            if i == j:
                continue
            dists_real.append(_row_distance(x, y, num_cols, cat_cols, scaler, distance))
        dists_real.sort()
        if len(dists_real) == 0:
            d_k_real = np.inf
        else:
            k_eff = min(k, len(dists_real))
            d_k_real = dists_real[k_eff - 1]
        # distances to synthetic
        dists_syn = []
        for _, z in synth_df.iterrows():
            dists_syn.append(_row_distance(x, z, num_cols, cat_cols, scaler, distance))
        dists_syn.sort()
        if len(dists_syn) == 0:
            d_k_syn = np.inf
        else:
            k_eff_s = min(k, len(dists_syn))
            d_k_syn = dists_syn[k_eff_s - 1]
        d_real_k_list.append(d_k_real)
        d_synth_k_list.append(d_k_syn)
        r = (d_k_real + eps) / (d_k_syn + eps)
        ratios.append(r)
    out = pd.DataFrame({
        'ratio': ratios,
        'd_k_real': d_real_k_list,
        'd_k_synth': d_synth_k_list
    })
    out.index.name = 'real_index'
    return out

# ----------------------------
# Toy Adult-like dataset
# ----------------------------
rng = np.random.default_rng(42)

def make_adult_like(n, age_mu=38, age_sd=10, hrs_mu=40, hrs_sd=10, workclass_p=(0.7, 0.2, 0.1)):
    ages = np.clip(rng.normal(age_mu, age_sd, n).round().astype(int), 18, 65)
    hours = np.clip(rng.normal(hrs_mu, hrs_sd, n).round().astype(int), 1, 99)
    workclasses = rng.choice(['Private', 'Self-emp', 'Govt'], size=n, p=workclass_p)
    df = pd.DataFrame({'age': ages, 'hours_per_week': hours, 'workclass': workclasses})
    df['workclass'] = df['workclass'].astype('category')
    # introduce a few NaNs to show handling
    if n >= 30:
        df.loc[::29, 'workclass'] = np.nan
    return df

# Members (train) and non-members (holdout) from similar distributions
real_train = make_adult_like(60, age_mu=39, hrs_mu=41)   # slight params
real_holdout = make_adult_like(60, age_mu=39, hrs_mu=41) # same params to simulate ideal holdout

# Synthetic datasets:
# 1) clone-like: contains many exact copies of train plus tiny perturbations
clone_indices = rng.choice(len(real_train), size=50, replace=True)
synth_clone = real_train.iloc[clone_indices].reset_index(drop=True)
# jitter some numerics slightly (but keep many exact)
jitter_idx = rng.choice(len(synth_clone), size=20, replace=False)
for col, scale in [('age', 1), ('hours_per_week', 1)]:
    synth_clone.loc[jitter_idx, col] = np.clip(synth_clone.loc[jitter_idx, col] + rng.integers(-1, 2, size=len(jitter_idx)), 0, 100)

# 2) safe-like: draws fresh from a slightly smoothed distribution (no direct copies)
synth_safe = make_adult_like(60, age_mu=39, hrs_mu=41)

# ----------------------------
# Worked examples
# ----------------------------

print("=== Example: dcr_membership_score ===")
# Pick 3 candidate rows: exact duplicate of a train row, a near-duplicate, and an outlier
candidate_exact = real_train.iloc[[0]].copy()
candidate_near = candidate_exact.copy()
candidate_near['age'] = candidate_near['age'] + 1  # small perturbation
candidate_outlier = pd.DataFrame({'age': [65], 'hours_per_week': [99], 'workclass': pd.Categorical(['Self-emp'])})

cands = pd.concat([candidate_exact, candidate_near, candidate_outlier], ignore_index=True)
dcr_gower = dcr_membership_score(real_train, cands, distance='gower', k=1)
dcr_heom = dcr_membership_score(real_train, cands, distance='heom', k=1)
print("Candidates:\n", cands)
print("DCR (Gower):", dcr_gower)
print("DCR (HEOM): ", dcr_heom)
print("Interpretation: lower values indicate closer to some training record; 0.0 means exact copy.")

print("\n=== Example: k_nn_mia_score (k=5, mean & k-th) ===")
knn_mean = k_nn_mia_score(real_train, synth_clone, distance='gower', k=5, agg='mean')
knn_kth  = k_nn_mia_score(real_train, synth_clone, distance='gower', k=5, agg='max')
print("Clone-like synth: k-NN mean distance (first 10):\n", knn_mean.head(10).to_string(index=False))
print("Clone-like synth: k-NN k-th distance (first 10):\n", knn_kth.head(10).to_string(index=False))
print("Summary (clone-like): mean={:.3f}, median={:.3f}".format(knn_mean.mean(), knn_mean.median()))

knn_mean_safe = k_nn_mia_score(real_train, synth_safe, distance='gower', k=5, agg='mean')
print("Safe-like synth: k-NN mean distance summary: mean={:.3f}, median={:.3f}".format(knn_mean_safe.mean(), knn_mean_safe.median()))
print("Interpretation: lower scores => synthetic samples lie very close to the training set (higher leakage risk).")

print("\n=== Example: nn_classifier_mia (1-NN on pooled real vs holdout) ===")
nn_result = nn_classifier_mia(real_train, real_holdout, distance='gower', protocol='holdout')
print("Accuracy: {:.3f}, AUC (pairwise): {:.3f}".format(nn_result['accuracy'], nn_result['auc']))
print("Head of per-row details:\n", nn_result['details'].head(5).to_string(index=False))
print("Interpretation: accuracy/AUC close to 0.5 => no membership advantage; higher => attack can tell train from holdout.")

print("\n=== Example: local_density_ratio (k=5) ===")
ldr_clone = local_density_ratio(real_train, synth_clone, k=5, distance='gower')
ldr_safe  = local_density_ratio(real_train, synth_safe,  k=5, distance='gower')
print("Clone-like: median ratio={:.3f}, fraction<1={:.2f}".format(ldr_clone['ratio'].median(), (ldr_clone['ratio']<1).mean()))
print("Safe-like : median ratio={:.3f}, fraction<1={:.2f}".format(ldr_safe['ratio'].median(),  (ldr_safe['ratio']<1).mean()))
print("Interpretation: ratio<1 => synthetic is denser around that real record (potential overfitting); ratio>1 => under-coverage.")
```

```python
# Fix the candidate selection to avoid NaN in workclass and correct the local_density_ratio interpretation.
import numpy as np
import pandas as pd
from math import sqrt

# Reuse earlier defined functions if present; otherwise redefine (safe redefinition).
def _split_cols(df: pd.DataFrame):
    num_cols = df.select_dtypes(include=['number']).columns.tolist()
    cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    for c in df.select_dtypes(include=['bool']).columns:
        if c not in cat_cols:
            cat_cols.append(c)
    return num_cols, cat_cols

def _get_scaler(df: pd.DataFrame, num_cols):
    return {c: (float(df[c].min()), float(df[c].max())) for c in num_cols}

def _row_distance(row1, row2, num_cols, cat_cols, scaler, metric: str):
    metric = metric.lower()
    if metric not in ('gower', 'heom', 'euclidean'):
        raise ValueError("Unsupported metric; choose 'gower', 'heom', or 'euclidean'")
    if metric == 'heom':
        s = 0.0
        for c in num_cols:
            lo, hi = scaler[c]
            rng = hi - lo if hi != lo else 1.0
            x = row1[c]; y = row2[c]
            if pd.isna(x) or pd.isna(y):
                d = 1.0
            else:
                d = (float(x) - float(y)) / rng
            s += d * d
        for c in cat_cols:
            x = row1[c]; y = row2[c]
            diff = 1.0 if (pd.isna(x) or pd.isna(y) or x != y) else 0.0
            s += diff * diff
        return sqrt(s)
    else:
        s = 0.0
        cnt = 0
        for c in num_cols:
            lo, hi = scaler[c]
            rng = hi - lo if hi != lo else 1.0
            x = row1[c]; y = row2[c]
            if pd.isna(x) or pd.isna(y):
                d = 1.0
            else:
                d = abs(float(x) - float(y)) / rng
            if metric == 'euclidean':
                s += d * d
            else:
                s += d
            cnt += 1
        for c in cat_cols:
            x = row1[c]; y = row2[c]
            diff = 1.0 if (pd.isna(x) or pd.isna(y) or x != y) else 0.0
            if metric == 'euclidean':
                s += diff * diff
            else:
                s += diff
            cnt += 1
        return sqrt(s) if metric == 'euclidean' else (s / (cnt if cnt > 0 else 1.0))

def dcr_membership_score(real_df: pd.DataFrame, candidate_rows: pd.DataFrame, 
                         distance: str = 'gower', scaler=None, k: int = 1):
    num_cols, cat_cols = _split_cols(real_df)
    if scaler is None:
        scaler = _get_scaler(real_df, num_cols)
    real_df = real_df.reset_index(drop=True)
    cand = candidate_rows.reset_index(drop=True)
    out = []
    for i, r in cand.iterrows():
        dists = []
        for j, rr in real_df.iterrows():
            d = _row_distance(r, rr, num_cols, cat_cols, scaler, distance)
            dists.append(d)
        dists.sort()
        k_eff = min(k, len(dists))
        out.append(float(np.mean(dists[:k_eff])) if k_eff > 0 else np.nan)
    return np.array(out)

def k_nn_mia_score(real_df: pd.DataFrame, synth_df: pd.DataFrame, 
                   distance: str = 'gower', k: int = 5, agg: str = 'mean'):
    num_cols, cat_cols = _split_cols(real_df)
    scaler = _get_scaler(real_df, num_cols)
    real_df = real_df.reset_index(drop=True)
    synth_df = synth_df.reset_index(drop=True)
    scores = []
    for i, s in synth_df.iterrows():
        dists = []
        for j, rr in real_df.iterrows():
            dists.append(_row_distance(s, rr, num_cols, cat_cols, scaler, distance))
        dists.sort()
        if len(dists) == 0:
            scores.append(np.nan)
        else:
            k_eff = min(max(k,1), len(dists))
            if agg == 'mean':
                scores.append(float(np.mean(dists[:k_eff])))
            elif agg == 'max':
                scores.append(float(dists[k_eff-1]))
            else:
                raise ValueError("agg must be 'mean' or 'max'")
    return pd.Series(scores, index=synth_df.index, name=f'{k}-NN_{agg}_{distance}')

def nn_classifier_mia(real_df: pd.DataFrame, holdout_df: pd.DataFrame, distance: str = 'gower', protocol: str = 'holdout'):
    assert protocol in ('holdout', 'pooled-loo')
    A = real_df.copy().reset_index(drop=True)
    B = holdout_df.copy().reset_index(drop=True)
    A['__label__'] = 1
    B['__label__'] = 0
    combined = pd.concat([A, B], ignore_index=True)
    num_cols, cat_cols = _split_cols(combined.drop(columns=['__label__']))
    scaler = _get_scaler(combined, num_cols)
    nearest_member = []
    nearest_nonmem = []
    pred = []
    for i, x in combined.iterrows():
        d_min_mem = np.inf
        d_min_non = np.inf
        for j, y in combined.iterrows():
            if i == j:
                continue
            d = _row_distance(x, y, num_cols, cat_cols, scaler, distance)
            if y['__label__'] == 1 and d < d_min_mem:
                d_min_mem = d
            if y['__label__'] == 0 and d < d_min_non:
                d_min_non = d
        nearest_member.append(d_min_mem)
        nearest_nonmem.append(d_min_non)
        pred.append(1 if d_min_mem <= d_min_non else 0)
    combined['d_nn_member'] = nearest_member
    combined['d_nn_nonmember'] = nearest_nonmem
    combined['pred'] = pred
    y_true = combined['__label__'].values
    y_pred = combined['pred'].values
    acc = float(np.mean(y_true == y_pred))
    # scores: larger -> more likely member
    scores = (combined['d_nn_nonmember'] - combined['d_nn_member']).values
    pos_scores = scores[y_true == 1]
    neg_scores = scores[y_true == 0]
    if len(pos_scores) == 0 or len(neg_scores) == 0:
        auc = np.nan
    else:
        wins = 0.0; total = 0.0
        for ps in pos_scores:
            for ns in neg_scores:
                wins += 1.0 if ps > ns else (0.5 if ps == ns else 0.0)
                total += 1.0
        auc = float(wins/total)
    return {'accuracy': acc, 'auc': auc, 'details': combined.drop(columns=['__label__'])}

def local_density_ratio(real_df: pd.DataFrame, synth_df: pd.DataFrame, k: int = 5, distance: str = 'gower'):
    """
    ratio_i = (d_k_real(i) + eps) / (d_k_synth(i) + eps)
    Interpretation:
      ratio > 1  -> synthetic is denser around i than real (potential overfitting risk).
      ratio ~ 1  -> similar local density: neutral.
      ratio < 1  -> real is denser than synthetic (under-coverage; possible utility issue).
    """
    eps = 1e-9
    real_df = real_df.reset_index(drop=True)
    synth_df = synth_df.reset_index(drop=True)
    num_cols, cat_cols = _split_cols(real_df)
    scaler = _get_scaler(real_df, num_cols)
    ratios = []
    d_real_k_list = []
    d_synth_k_list = []
    for i, x in real_df.iterrows():
        dists_real = []
        for j, y in real_df.iterrows():
            if i == j: 
                continue
            dists_real.append(_row_distance(x, y, num_cols, cat_cols, scaler, distance))
        dists_real.sort()
        d_k_real = dists_real[min(k-1, max(0, len(dists_real)-1))] if len(dists_real) else np.inf
        dists_syn = []
        for _, z in synth_df.iterrows():
            dists_syn.append(_row_distance(x, z, num_cols, cat_cols, scaler, distance))
        dists_syn.sort()
        d_k_syn = dists_syn[min(k-1, max(0, len(dists_syn)-1))] if len(dists_syn) else np.inf
        d_real_k_list.append(d_k_real)
        d_synth_k_list.append(d_k_syn)
        ratios.append((d_k_real + eps) / (d_k_syn + eps))
    return pd.DataFrame({'ratio': ratios, 'd_k_real': d_real_k_list, 'd_k_synth': d_synth_k_list})

# Rebuild toy data like before (same RNG seed for reproducibility)
rng = np.random.default_rng(42)
def make_adult_like(n, age_mu=38, age_sd=10, hrs_mu=40, hrs_sd=10, workclass_p=(0.7, 0.2, 0.1)):
    ages = np.clip(rng.normal(age_mu, age_sd, n).round().astype(int), 18, 65)
    hours = np.clip(rng.normal(hrs_mu, hrs_sd, n).round().astype(int), 1, 99)
    workclasses = rng.choice(['Private', 'Self-emp', 'Govt'], size=n, p=workclass_p)
    df = pd.DataFrame({'age': ages, 'hours_per_week': hours, 'workclass': workclasses})
    df['workclass'] = df['workclass'].astype('category')
    if n >= 30:
        df.loc[::29, 'workclass'] = np.nan
    return df

real_train = make_adult_like(60, age_mu=39, hrs_mu=41)
real_holdout = make_adult_like(60, age_mu=39, hrs_mu=41)

# synth clone and safe as before
clone_indices = rng.choice(len(real_train), size=50, replace=True)
synth_clone = real_train.iloc[clone_indices].reset_index(drop=True)
jitter_idx = rng.choice(len(synth_clone), size=20, replace=False)
for col in ['age','hours_per_week']:
    synth_clone.loc[jitter_idx, col] = np.clip(synth_clone.loc[jitter_idx, col] + rng.integers(-1, 2, size=len(jitter_idx)), 0, 100)

synth_safe = make_adult_like(60, age_mu=39, hrs_mu=41)

print("=== Example: dcr_membership_score (fixed candidates) ===")
# pick a train row with non-missing workclass
idx_non_nan = int(np.where(~real_train['workclass'].isna())[0][0])
candidate_exact = real_train.iloc[[idx_non_nan]].copy()
candidate_near = candidate_exact.copy()
candidate_near['age'] = candidate_near['age'] + 1
candidate_outlier = pd.DataFrame({'age': [65], 'hours_per_week': [99], 'workclass': pd.Categorical(['Self-emp'])})
cands = pd.concat([candidate_exact, candidate_near, candidate_outlier], ignore_index=True)
dcr_gower = dcr_membership_score(real_train, cands, distance='gower', k=1)
dcr_heom  = dcr_membership_score(real_train, cands, distance='heom',  k=1)
print("Candidates:\n", cands.to_string(index=False))
print("DCR (Gower):", np.round(dcr_gower, 6))
print("DCR (HEOM) :", np.round(dcr_heom, 6))

print("\n=== Example: k_nn_mia_score (k=5, mean & k-th) ===")
knn_mean_clone = k_nn_mia_score(real_train, synth_clone, distance='gower', k=5, agg='mean')
knn_kth_clone  = k_nn_mia_score(real_train, synth_clone, distance='gower', k=5, agg='max')
print("Clone-like synth: mean of 5-NN distances: mean={:.3f}, median={:.3f}".format(knn_mean_clone.mean(), knn_mean_clone.median()))
print("Clone-like synth: k-th (5th) distances:   mean={:.3f}, median={:.3f}".format(knn_kth_clone.mean(), knn_kth_clone.median()))
knn_mean_safe = k_nn_mia_score(real_train, synth_safe, distance='gower', k=5, agg='mean')
print("Safe-like synth: mean of 5-NN distances:  mean={:.3f}, median={:.3f}".format(knn_mean_safe.mean(), knn_mean_safe.median()))

print("\n=== Example: nn_classifier_mia (1-NN on pooled real vs holdout) ===")
nn_result = nn_classifier_mia(real_train, real_holdout, distance='gower', protocol='holdout')
print("Accuracy: {:.3f}, AUC (pairwise): {:.3f}".format(nn_result['accuracy'], nn_result['auc']))
print(nn_result['details'].head(5).to_string(index=False))

print("\n=== Example: local_density_ratio (k=1 and k=5) ===")
ldr1_clone = local_density_ratio(real_train, synth_clone, k=1, distance='gower')
ldr5_clone = local_density_ratio(real_train, synth_clone, k=5, distance='gower')
ldr1_safe  = local_density_ratio(real_train, synth_safe,  k=1, distance='gower')
ldr5_safe  = local_density_ratio(real_train, synth_safe,  k=5, distance='gower')
def summarize(name, df):
    return f"{name}: median={df['ratio'].median():.3f}, frac>1={ (df['ratio']>1).mean():.2f}, frac≈1 (±5%)={ ((df['ratio']>0.95)&(df['ratio']<1.05)).mean():.2f}"
print(summarize("Clone k=1", ldr1_clone))
print(summarize("Clone k=5", ldr5_clone))
print(summarize("Safe  k=1", ldr1_safe))
print(summarize("Safe  k=5", ldr5_safe))
print("Interpretation: ratio>1 => synthetic is locally denser (risk); ratio≈1 neutral; ratio<1 under-coverage by synthetic.")
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import NearestNeighbors


synth_data = credit_ctgan['generation_results']['synthetic_data']
synth_data_0005 = credit_ctgan_heom_any_eps_0005['generation_results']['synthetic_data']
# ---- 0.1 Cast dtypes once ----
for c in cat_list_credit_card:
    # keep as string to preserve unseen values across splits, but mark NaNs
    df_real_credit_card_train[c] = df_real_credit_card_train[c].astype('object')
    df_real_credit_card_test[c]  = df_real_credit_card_test[c].astype('object')
    synth_data[c]              = synth_data[c].astype('object')
    synth_data_0005[c]         = synth_data_0005[c].astype('object')

for c in num_list_credit_card:
    df_real_credit_card_train[c] = pd.to_numeric(df_real_credit_card_train[c], errors='coerce')
    df_real_credit_card_test[c]  = pd.to_numeric(df_real_credit_card_test[c], errors='coerce')
    synth_data[c]              = pd.to_numeric(synth_data[c], errors='coerce')
    synth_data_0005[c]         = pd.to_numeric(synth_data_0005[c], errors='coerce')
    

# ---- 0.2 Encoder that supports HEOM (Euclidean) and Gower (Manhattan) ----
class TabularEncoder:
    def __init__(self, num_cols, cat_cols):
        self.num_cols = list(num_cols)
        self.cat_cols = list(cat_cols)
        self.num_scaler = MinMaxScaler()
        self.cat_levels = {}   # per-col list of levels (incl '__nan__')
        self.cat_col_slices = {}  # per-col slice indices into the OHE block
        self.ohe_cols = []     # names of OHE columns in order

    def _prep_cats(self, df):
        # map to strings, NaN -> '__nan__'; unknown -> '__nan__'
        data = {}
        for c in self.cat_cols:
            s = df[c].astype(str)
            s = s.where(~s.isna(), '__nan__')
            # map unknown categories to '__nan__'
            known = set(self.cat_levels[c])
            s = s.where(s.isin(known), '__nan__')
            data[c] = s
        return pd.DataFrame(data)

    def fit(self, df_fit):
        # numeric scaler
        self.num_scaler.fit(df_fit[self.num_cols])

        # categorical vocab learned on train only (privacy‑conservative)
        for c in self.cat_cols:
            s = df_fit[c].astype(str)
            s = s.where(~s.isna(), '__nan__')
            levels = sorted(s.unique().tolist())
            if '__nan__' not in levels:
                levels.append('__nan__')
            self.cat_levels[c] = levels

        # build OHE column order and per‑col slices
        self.ohe_cols = []
        start = 0
        for c in self.cat_cols:
            levs = self.cat_levels[c]
            self.ohe_cols += [f"{c}=={v}" for v in levs]
            end = start + len(levs)
            self.cat_col_slices[c] = slice(start, end)
            start = end
        return self

    def transform_HEOM(self, df):
        # numerics in [0,1]
        X_num = self.num_scaler.transform(df[self.num_cols]).astype('float32')  # shape (n, p_num)

        # categorical OHE with __nan__ as a legit level
        C = self._prep_cats(df)
        blocks = []
        for c in self.cat_cols:
            levs = self.cat_levels[c]
            one = pd.Categorical(C[c], categories=levs)
            ohe = pd.get_dummies(one, prefix=f"{c}==", dtype=np.float32)
            # ensure all columns present in the right order
            missing = [f"{c}=={v}" for v in levs if f"{c}=={v}" not in ohe.columns]
            for m in missing: ohe[m] = 0.0
            ohe = ohe[[f"{c}=={v}" for v in levs]]
            blocks.append(ohe.values)
        X_cat = np.concatenate(blocks, axis=1) if blocks else np.zeros((len(df), 0), dtype=np.float32)

        # scale categorical block so that a mismatch contributes EXACTLY 1 in L2:
        # with one‑hot, L2 mismatch => sqrt(2); multiply all cat dims by 1/sqrt(2)
        if X_cat.shape[1] > 0:
            X_cat *= (1.0 / np.sqrt(2.0)).astype(np.float32)

        return np.hstack([X_num, X_cat]).astype('float32')

    def transform_GOWER(self, df):
        # numerics in [0,1]
        X_num = self.num_scaler.transform(df[self.num_cols]).astype('float32')

        # categorical one‑hot (as above)
        C = self._prep_cats(df)
        blocks = []
        for c in self.cat_cols:
            levs = self.cat_levels[c]
            one = pd.Categorical(C[c], categories=levs)
            ohe = pd.get_dummies(one, prefix=f"{c}==", dtype=np.float32)
            missing = [f"{c}=={v}" for v in levs if f"{c}=={v}" not in ohe.columns]
            for m in missing: ohe[m] = 0.0
            ohe = ohe[[f"{c}=={v}" for v in levs]]
            blocks.append(ohe.values)
        X_cat = np.concatenate(blocks, axis=1) if blocks else np.zeros((len(df), 0), dtype=np.float32)

        # L1 mismatch across one‑hot is 2; scale by 1/2 so a mismatch contributes 1 in L1
        if X_cat.shape[1] > 0:
            X_cat *= 0.5

        X = np.hstack([X_num, X_cat]).astype('float32')
        # For Gower value (not NN ranking), divide by feature count (num features + cat features as 1 each)
        self._gower_feature_count = len(self.num_cols) + len(self.cat_cols)
        return X

# ---- 0.3 Fit on train and transform all splits ----
enc = TabularEncoder(num_cols=num_list_credit_card, cat_cols=cat_list_credit_card).fit(df_real_credit_card_train)


X_train_heom = enc.transform_HEOM(df_real_credit_card_train)
X_hold_heom  = enc.transform_HEOM(df_real_credit_card_test)
X_syn_heom   = enc.transform_HEOM(synth_data)
X_syn_0005_heom   = enc.transform_HEOM(synth_data_0005)

X_train_gow = enc.transform_GOWER(df_real_credit_card_train)
X_hold_gow  = enc.transform_GOWER(df_real_credit_card_test)
X_syn_gow   = enc.transform_GOWER(synth_data)
X_syn_0005_gow   = enc.transform_GOWER(synth_data_0005)
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

# --- build NN over train (HEOM) ---
nn_train_heom = NearestNeighbors(n_neighbors=2, metric='euclidean').fit(X_train_heom)

# DCR for holdout → train
d_hold_to_train = nn_train_heom.kneighbors(X_hold_heom, n_neighbors=1, return_distance=True)[0].ravel()

# DCR for train (LOO) → train (skip self by taking 2-NN and drop the first)
d_train_loo = nn_train_heom.kneighbors(X_train_heom, n_neighbors=2, return_distance=True)[0][:, 1]

# Membership scoring: smaller distance = more likely member
labels = np.r_[np.ones(len(d_train_loo)), np.zeros(len(d_hold_to_train))]
scores  = -np.r_[d_train_loo, d_hold_to_train]   # negate so larger=more member for ROC

auc = roc_auc_score(labels, scores)
fpr, tpr, thr = roc_curve(labels, scores)
# TPR at ~1% FPR:
target_fpr = 0.01
i = np.searchsorted(fpr, target_fpr)
tpr_at_1pct = tpr[i] if i < len(tpr) else tpr[-1]

print(f"[HEOM] DCR threshold MIA: AUC={auc:.3f}, TPR@1%FPR={tpr_at_1pct:.3f}")

# Repeat for Gower (use Manhattan; divide by feature count to get Gower in [0,1] if you need the value)
nn_train_gow = NearestNeighbors(n_neighbors=2, metric='manhattan').fit(X_train_gow)
d_hold_to_train_g = nn_train_gow.kneighbors(X_hold_gow, n_neighbors=1, return_distance=True)[0].ravel() / enc._gower_feature_count
d_train_loo_g     = nn_train_gow.kneighbors(X_train_gow, n_neighbors=2, return_distance=True)[0][:, 1] / enc._gower_feature_count

labels_g = np.r_[np.ones(len(d_train_loo_g)), np.zeros(len(d_hold_to_train_g))]
scores_g = -np.r_[d_train_loo_g, d_hold_to_train_g]
auc_g = roc_auc_score(labels_g, scores_g)
fpr_g, tpr_g, _ = roc_curve(labels_g, scores_g)
i = np.searchsorted(fpr_g, target_fpr)
tpr_at_1pct_g = tpr_g[i] if i < len(tpr_g) else tpr_g[-1]
print(f"[Gower] DCR threshold MIA: AUC={auc_g:.3f}, TPR@1%FPR={tpr_at_1pct_g:.3f}")
```
